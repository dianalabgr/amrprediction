```python
import xgboost as xgb
import numpy as np
import pickle
import os
import pandas as pd
import numpy as np
from scipy.sparse import csr_matrix
```

```python
base_dir = "./models_and_ShapValues/"

models = {}
# Walk through the directory structure
for root, dirs, files in os.walk(base_dir):
    # Check if we're in a 'models' directory
    print(root)
    print(dirs)
    if root.endswith('models'):
        for file in files:
            if file.endswith('.pkl'):
                # Extract antibiotic name from filename
                print(file)
                if '_xgboost' in file:
                    antibiotic = file.split('_xgboost')[0]
                    model_type = 'XGBoost'
                elif '_RandomForest' in file:
                    antibiotic = file.split('_RandomForest')[0]
                    model_type = 'RandomForest'
                elif '_random_forest' in file:
                    antibiotic = file.split('_RandomForest')[0]
                    model_type = 'RandomForest'
                print(antibiotic)
                # Extract kmer size (assuming format like "kmer_3" or "kmer_5")
                kmer_size = int(file.split('kmer_')[1].split('.pkl')[0])
                print(kmer_size)
                  # Load the model and store with kmer size
                with open(os.path.join(root, file), 'rb') as f:
                    models[antibiotic] = {  # Remove 'model_' prefix
                        'model': pickle.load(f),
                        'best_kmer': kmer_size,
                        'model_type': model_type
                    }

```

```python
def predict_amr(sample_name, genus_antibiotics):
    results = {'Antibiotic': [], 'Prediction': []}
    
    for antibiotic in genus_antibiotics:
        if antibiotic in models:
            kmer_size = models[antibiotic]['best_kmer']
            #the path where the kmers are saved from the annotation
            kmer_file = f"../annotate_newAssemblies/kmers_Escherichia/kmer_{kmer_size}_flattened_{sample_name}.npz"
            
            try:
                X_npz = np.load(kmer_file, allow_pickle=True)
                X_dna = X_npz['all_kmer_counts_dna'].reshape(1, -1)
                X_rrna = X_npz['all_kmer_counts_rrna'].reshape(1, -1)
                X_protein = X_npz['all_kmer_counts_protein'].reshape(1, -1)
                X = np.concatenate((X_dna, X_rrna, X_protein), axis=1)
                X = csr_matrix(X)
            except FileNotFoundError:
                results['Antibiotic'].append(antibiotic)
                results['Prediction'].append('File Not Found')
                continue
            
            model = models[antibiotic]['model']

            if isinstance(model, xgb.Booster):
                dtest = xgb.DMatrix(X)
                y_pred = model.predict(dtest)
                prediction = "R" if y_pred[0] > 0.5 else "S"
            else:  # XGBClassifier or RandomForest
                y_pred = model.predict(X)
                prediction = "R" if y_pred[0] > 0.5 else "S"

        else:
            prediction = 'NA'
        
        results['Antibiotic'].append(antibiotic)
        results['Prediction'].append(prediction)
    
    return pd.DataFrame(results)
```
